```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def sigmoid_derivative(x):
    s = sigmoid(x)
    return s * (1 - s)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

z = np.linspace(-10, 10, 400)

sigmoid_output = sigmoid(z)
relu_output = relu(z)

sigmoid_grad = sigmoid_derivative(z)
relu_grad = relu_derivative(z)

sample_values = np.array([-5, -2, -1, 0, 1, 2, 5])

comparison_table = {
    "z": sample_values,
    "sigmoid(z)": np.round(sigmoid(sample_values), 4),
    "relu(z)": np.round(relu(sample_values), 4),
    "sigmoid_derivative(z)": np.round(sigmoid_derivative(sample_values), 4),
    "relu_derivative(z)": np.round(relu_derivative(sample_values), 4)
}

print("Activation Function Comparison Table")
for i in range(len(sample_values)):
    print(
        f"z={comparison_table['z'][i]:>3} | "
        f"sigmoid={comparison_table['sigmoid(z)'][i]:>6} | "
        f"relu={comparison_table['relu(z)'][i]:>4} | "
        f"sigmoid'={comparison_table['sigmoid_derivative(z)'][i]:>6} | "
        f"relu'={comparison_table['relu_derivative(z)'][i]:>2}"
    )

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(z, sigmoid_output, label="Sigmoid")
plt.plot(z, relu_output, label="ReLU")
plt.title("Activation Functions")
plt.xlabel("Linear Output (z)")
plt.ylabel("Activated Output")
plt.axhline(0, color="black", lw=0.5)
plt.axvline(0, color="black", lw=0.5)
plt.grid(True)
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(z, sigmoid_grad, label="Sigmoid Derivative")
plt.plot(z, relu_grad, label="ReLU Derivative")
plt.title("Activation Function Derivatives")
plt.xlabel("Linear Output (z)")
plt.ylabel("Gradient")
plt.axhline(0, color="black", lw=0.5)
plt.axvline(0, color="black", lw=0.5)
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()

print("\nKey Understanding")
print("1. z is the raw output from a neuron before activation.")
print("2. Sigmoid converts values to the range 0 to 1, useful for probability-like outputs.")
print("3. ReLU returns 0 for negative values and keeps positive values unchanged.")
print("4. Derivatives show how learning happens during backpropagation.")
print("5. Sigmoid can saturate for large positive or negative values.")
print("6. ReLU helps deep networks learn faster in many hidden-layer cases.")
```
